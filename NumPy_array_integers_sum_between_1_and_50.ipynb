```python
import numpy as np

arr = np.random.randint(1, 51, size=(5, 5))
print("Original array:\n", arr)

row_sums = arr.sum(axis=1)
max_row_index = np.argmax(row_sums)

print("\nRow sums:", row_sums)
print("Row with maximum sum:", arr[max_row_index])

# Replace all elements greater than the mean with 0
mean_value = arr.mean()
modified_arr = arr.copy()
modified_arr[modified_arr > mean_value] = 0

print("\nMean:", mean_value)
print("Modified array:\n", modified_arr)
```
